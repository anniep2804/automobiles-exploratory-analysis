#### Which are the 5 most expensive cars?

How do the most expensive and cheapest cars compare? Exploring the most expensive cars highlights if some moviecars are worth the money spent on them based on their fuel economy (mpg or miles per gallon)

```python
# create histogram for distribution of car prices
plt.figure(figsize=(12, 6))
sns.histplot(automobiles_df['price'], bins=20, kde=True, color='red')
plt.title('Distribution of Car Prices')
plt.xlabel('Price')
plt.ylabel('Count')
plt.show()
```

```python
# get rows for 5 most expensive and 5 cheapest cars 
highest_price = automobiles_df.sort_values(by='price', ascending=False).head(5)
lowest_price = automobiles_df.sort_values(by='price', ascending=False).tail(5)
print("5 Most Expensive Cars")
print(highest_price[['make', 'price']])
print("5 Cheapest Cars")
print(lowest_price[['make', 'price']])

# get average mpg for city and highway 
automobiles_df['average-mpg'] = (automobiles_df['city-mpg'] + automobiles_df['highway-mpg']) / 2

# analysis of mpg vs price for the 5 most expensive cars
plt.figure(figsize=(12, 6))
plt.scatter(highest_price['average-mpg'], highest_price['price'], alpha=0.5, color='green')
plt.xlabel('MPG')
plt.ylabel('Price')
plt.title('MPG vs Price for 5 Most Expensive Cars')
plt.show()

# analysis of mpg vs price for the 5 least expensive cars
plt.figure(figsize=(12, 6))
plt.scatter(lowest_price['average-mpg'], lowest_price['price'], alpha=0.5, color='red')
plt.xlabel('MPG')
plt.ylabel('Price')
plt.title('MPG vs Price for 5 Cheapest Cars')
plt.show()

# anaylsys of horsepower vs price for 5 most expensive cars 
plt.figure(figsize=(12, 6))
plt.scatter(highest_price['horsepower'], highest_price['price'], alpha=0.5, color='yellow')
plt.xlabel('Horsepower')
plt.ylabel('Price')
plt.title('Horsepower vs Price for 5 Most Expensive Cars')
plt.show()

# analysis of horsepower vs price for 5 cheapest cars 
plt.figure(figsize=(12, 6))
plt.scatter(lowest_price['horsepower'], lowest_price['price'], alpha=0.5, color='blue')
plt.xlabel('Horsepower')
plt.ylabel('Price')
plt.title('Horsepower vs Price for 5 Cheapest Cars')
plt.show()
```

#### Which manufacturer builds the most fuel efficient vehicles?

Compare the average mpg for each vehicle manufacture's vehicles and create a bar plot
